This notebook demonstrates the reusable Python implementation. Download FloodNet separately, update the two paths below, and cite [Safavi & Rahnemoonfar (2023)](https://doi.org/10.1109/JSTARS.2022.3219724) when using this work.

````python
from floodnet_segformer import (
    FloodNetDataset, TrainConfig, build_model, discover_ids, split_ids, train
)

IMAGE_DIR = '/path/to/FloodNet/images'
MASK_DIR = '/path/to/FloodNet/masks'
````

## Inspect the reproducible data split

````python
train_ids, val_ids, test_ids = split_ids(discover_ids(IMAGE_DIR), seed=19)
print({'train': len(train_ids), 'validation': len(val_ids), 'test': len(test_ids)})
sample_set = FloodNetDataset(IMAGE_DIR, MASK_DIR, train_ids[:2], augment=True)
image, mask = sample_set[0]
print('image:', tuple(image.shape), 'mask:', tuple(mask.shape))
````

## Train
Change `nvidia/mit-b3` to `nvidia/mit-b0` for the smaller variant. Start with one epoch to verify the setup, then increase to 15 for the paper-style experiment.

````python
config = TrainConfig(
    image_dir=IMAGE_DIR,
    mask_dir=MASK_DIR,
    model_name='nvidia/mit-b3',
    output_dir='outputs/segformer-b3',
    epochs=1,  # use 15 for the full run
    batch_size=1,
)
history = train(config)
````
